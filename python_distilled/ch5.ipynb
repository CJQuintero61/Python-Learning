```python
"""Default values are assigned by assigning them in the function definintion"""
def split(line, delimiter=','):
    return line.split(delimiter)

"""
NOTE When a parameter has a default value, it is considered optional. That argument and all
following args must have default values and are all optional.

NOTE default parameter values are evaluated only once when the function is first defined, not
each time the function is called.
"""

# Bad
def func(x, items=[]):
    items.append(x)
    return items

func(1)  # returns [1]
func(2)  # returns [1, 2] because the default list is shared across calls
func(3)  # returns [1, 2, 3] because the default list is shared across calls

"""
For this reason, you should only use immutable objects as default values like
strings, numbers, booleans, None, and so on.
"""

# Better
def func(x, items=None):
    if items is None:
        items = []
    items.append(x)
    return items
```

```python
"""
Variadic Arguments

A function can accept a variable number of args if its last param has an asterisk before it.
In this case, all the extra args placed in that variable as a tuple which supports
standard sequence operations like unpacking, iteration, and slicing
"""
def product(first, *args):
    total = first
    for arg in args:
        total = total * arg
    return total

product(2, 3)  # returns 6
product(2, 3, 4, 5, 6)  # returns 720
```

```python
"""
Keyword Arguments

A function can be called with keyword arguments. In this case, order of args doesn't matter
as long as each required arg is provided.
"""

def func(a, b, c):
    return a + b + c

func(a = 1, b = 2, c = 3)  # returns 6

"""
You can also combine positional and keyword args as long as the positional args come first
"""
func('hello', 3, c=[1,2], y=22)

"""
You can force the use of keyword args by listing args after a * argument, or by just using a
* in the function definition.
"""

def read_data(filename, *, debug=True):
    print(f"Reading data from {filename}")

data = read_data('data.txt', debug=False)  # valid
data = read_data('data.txt', False)  # invalid, debug must be a keyword arg
```

```python
"""
Variadic Keyword Arguments (aka dictionary packing)

If the last argument of a function is prefixed with **, it will collect
any extra params and make a dictionary of them. This dictionary is guaranteed
to be in the order of the arguments passed to the function.

This is useful for functions that accept a lot of optional arguments
"""
def func(a, b, **kwargs):
    print(f"a: {a}, b: {b}, kwargs: {kwargs}")
```

```python
"""
Function Application and Parameter Passing

Python passes parameters 'as is' without any extra copying. This is especially important for
mutable objects like lists and dicts.
"""

def square(items):
    for i, x in enumerate(items):
        items[i] = x * x

# a gets modified in place
a = [1, 2, 3, 4, 5]
square(a)  # a is now [1, 4, 9, 16, 25]

print(a)  # prints [1, 4, 9, 16, 25]
"""
These problems are called side effects and are bad practice and should generally be avoided.

You can avoid this by updating local variable names bound to different objects
"""

def sum_square(items):
    # no side effect here since items is bound to the list comprehension and is a new object
    # you are reassigning the name to a different object, not modifying the original list in place
    items = [x * x for x in items]
    return sum(items)

a = [1, 2, 3, 4, 5]
sum_square(a)  # returns 55
print(a)  # prints [1, 2, 3, 4, 5] because a was not modified in place

"""
Stylistically, it is common for functions with side effects to return None
"""
items = [10, 3, 5, 7]
items.sort()  # sorts in place and returns None
items  # prints [3, 5, 7, 10]
```

```python
"""
Return Values

Things can be returned like other languages. If omitted, None is returned.

Sometimes you will want to return multiple things. To do so, return a tuple that
can be stored in a single var, or unpacked into many vars
"""
x = 1
y = 2

def return_a_tuple(x, y):
    x = x ** 2
    y = y ** 2
    return x, y

x_squared, y_squared = return_a_tuple(x, y)

"""
Sometimes named tuples are used as an alternative to treat things like
objects and attributes instead of a plain tuple
"""

from typing import NamedTuple

class ParseResult(NamedTuple):
    name: str
    value: str

def parse_value(text):
    """Split text of the form name=val into (name, value)"""
    parts = text.split('=', 1)
    return ParseResult(parts[0].strip(), parts[1].strip())

r = parse_value('url=https://www.python.org')
print(r.name, r.value)
```

```python
"""
Scoping Rules

There are 2 main scoping related errors: NameErrors and UnboundLocalError.

NameError - looking up an undefined name of a free var in the global environment

UnboundLocalError - referencing a local variable that isn't defined
"""

def func(x):
    if x > 0:
        y = 42
    return x + y

func(10)
# func(-10)   # UnboundLocalError

# UnboundLocalErrors are also common when updating a value that
# hasn't been initialized yet

def funky():
    n = n + 1
    return n

# funky() # UnboundLocalError

"""
Variables are either local or global in Python belonging to a specific namespace
This can also cause UnboundLocalErrors when you are working with globals at times
"""

x = 42
def funky_fresh():
    print(x) # UnboundLocalError
    x = 13

# funky_fresh()

# because x is defined in the local namespace of the function, the print fails and DOES NOT
# reference the global x variable as you may expect

def funky_and_fresh():
    x = 13 # does not overwrite the global x variable

funky_and_fresh()
print(x)

"""
If you need to modify a global (not recommended and un-pythonic)
you can use the global keyword

def func():
    global x # 'x' refers to the global variable x now
    x = 13

if you need to modify states behind the scenes, then use a class definition and modify state
by changing an instance or a class variable
"""

class Config():
    x = 42

def func():
    Config.x = 13

"""
Nested Functions


Python supports nested functions but this is not common and flat coding is better than nested coding.
It also makes it harder to debug and test nested functions. It can at times simplify complex logic
and hiding internal implementation details.
"""
```

```python
"""
Recursion

Python supports recursion and recursive functions and these are bound by the
sys.getrecursionlimit() and sys.setrecursionlimit() where the default is 1000.

You can increase the limit if needed, but it is still bound by the OS of the host system
and all the resources available.

If you are ever dealing with very large data structures like trees and graphs which are naturally recursive,
you may need to do some clever thinking with some clever operations.
"""

def factorial(x):
    if x == 1:
        return x
    else:
        return x * factorial(x - 1)

print(factorial(7))
```

```python
"""
Lambda

The notorious lambda function

Lambda is just an unnamed function in the format 'lambda args: expression'
where args is a comma separated list of args and the expression involves those args

Take caution when using lambda to reference variables that not bound to the lambda args list as late binding can occur
which causes problems at times.
"""

a = lambda x, y: x + y
r = a(2, 3)

print(r)
```

```python
"""
Higher Order Functions

Functions are first class objects, meaning they can be treated as any other object and can be passed 
as arguments to other functions, placed inside of data structures, and returned by other functions.
"""

# example of a callback function, because it 'calls back' the function supplied as an argument
import time

# call back function 
def after(seconds, func):
    time.sleep(seconds)
    func()

def greeting():
    print("Hello World")

after(5, greeting)  # prints Hello World after sleeping for 5
```
