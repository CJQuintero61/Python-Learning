Another common looping problem is iterating in parallel over 2 iterables such as:

```python
a = [x + 1 for x in range(10)]      # list from 1-10
b = [x + 1 for x in range(20)]      # list from 1-20

i = 0
while i < len(a) and i < len(b):
    x = a[i]    # take an item from list a
    y = b[i]    # take an item from list b
    i = i + 1

# the loop stops when either list runs out of elements
print(x)
print(y)
```

this can be simplified using zip() which 
combines iterables into tuples (a[0], b[0]), (a[1], b[1]), (a[2], b[2]) ...
and will stop when the shorter of the lists runs out of elements.

zip() returns a tuple that can be unpacked, or stored in a single variable

```python
a = [x + 1 for x in range(10)]      # list from 1-10
b = [x + 1 for x in range(20)]      # list from 1-20

for x, y in zip(a,b):
    print(x, y)
```

Defining new exceptions

to make one, define a class that inherits from Exception

When defining the class, the exceptions class constructor takes optional arguments.

You can make the constructor of your exception to take more params

```python
class DeviceError(Exception):
    def __init__(self, errno, msg):
        self.args = (errno, msg)
        self.errno = errno
        self.errmsg = msg

raise DeviceError(123456, 'Device Not Responding')
```

When defining an exception class that redefines __init__(), it is important to assign a
tuple with the args of __init__() to the attribute self.args as shown.
This attribute is used when printing exception traceback messages. 

If you leave that undefined, users won't be able to see any useful info about exceptions when they occur

## context managers and the with statement

the "with" statement lets statements execute inside a runtime context that is controlled by the context manager

with open(file, mode) as file:
    ...

this will auto close the file when execution leaves the with block

with lock:
    ...

this will auto acquire and release the lock when control enters and leaves the block

--

You can also use with obj to allow obj to manage what happens when control enters and exits the block

When "with obj" is called, obj.__enter__() is called to signal that a new context is being entered.

when control leaves, the obj.__exit__(type, value, traceback) is called

if no exception was raised, then the 3 args are set to None, otherwise they have the exception values.

if __exit__() returns True, it shows that the raised Exception was handled and should no longer be propagated. Returning None or False will cause
the error to propagate.

```python
class MyObject:
    def __enter__(self):
        print('entered the context manager')
    def __exit__(self, type, value, traceback):
        print('leaving the context manager')


with MyObject() as obj:
    # obj.a_function_you_defined()
    pass
```
