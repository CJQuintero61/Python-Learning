# Objects, Types, and Protocols

Everything in python is an object

```python
a = 42
"""
'a' is the label for the objects location in memory where the label isn't actually a part of the object itself.
42 is the value of the object that is being referenced by the label 'a'.

The 'identity' of an object is its memory address

Objects that hold a reference to another object are called 'containers'.
"""
```

```python
a + 10
"""
Under the hood, operators are mapped to special methods. In this case, the + operator is mapped
to the __add__ method so this is essentially calling a.__add__(10).
"""
```

```python
"""
Object Identity and Type

the built-in function id() returns the identity of an object. This is usually an integer that corresponds
to the object's location in memory.
"""

"""
The 'is' and 'is not' operators compare the identity of 2 objects. 

type() returns the type of an ojbect
"""
a = [1, 2, 3]
b = [1, 2, 3]

def compare(a, b):
    if a is b:
        print('same object')
    if a == b:
        print('same value')
    if type(a) is type(b):
        print('same type')

compare(a, a)
print()
compare(a, b)
print()
compare(a, [1, 2, 3])
```

```python
"""
We already know how to instatiate types.

Subtypes are defined by inheritance and add new features or redefine existing features of the parent type.
"""

# inherits the list type
class MyList(list):

    def remove_all(self, value):
        return [item for item in self if item != value]

items = MyList([5, 8, 2, 7, 2, 13, 9])
x = items.remove_all(2)
print(x)

"""
isinstance() is the preferred way to check since it knows of subtypes
"""
```

```python
"""
Reference Counting and Garbage Collection

All objects have a reference count that tracks how many references point to an object.
The count increases when a new ref is made by assigning the object to a new name (label) 
or placing it in a container.
"""

a = 37  # creates an object with the value 37
b = a   # creates a new reference to the same object (increases the reference count on 37)
c = []
c.append(a) # creates a new reference to the same object (increases the reference count on 37)

# in realtiy, this is 1 object with 3 references to it

"""
Objects reference counts are decreased by using 'del' or when a reference goes out of scope or is
reassigned to a new object
"""
del a # decreases the reference count on 37
b = 42 # decreases the reference count on 37 and increases the reference count on 42
c[0] = 2.0 # decreases the reference count on 37 and increases the reference count on 2.0

"""
You can use sys.getrefcount() on an object to get its reference count. This is often higher
than you expect because of behind the scenes stuff that python does.
"""


"""
You can invoke the cyclic garbage collector with the gc module.
gc.collect()

This is useful for very large objects and strucutres that are no longer needed
when you are wondering why your program is consuming so much memory.
"""
```

```python
"""
References and Copies

Strings and numbers are immutable, so they cannot be changed in place.

When using mutable things, references can be a problem
"""
a = [1, 2, 3, 4]
b = a # b is a reference to the same object as a

# they point to the same thing
print(b is a) # True

b[2] = -100 

print(a) # [1, 2, -100, 4]
```

```python
"""
Shallow copies

This is important so lock in!

There are 2 types of copies: shallow and deep.

A shallow copy creates a new object, but fills it with references to the items in the
original object.
"""
a = [1, 2, [3, 4]]
b = list(a) # creates a shallow copy of a

print(b is a) # False

b.append(100) # adds a new item to b, but not a
print(b) # [1, 2, [3, 4], 100]

"""Notice how a is not changed by the append statement on b"""
print (a) # [1, 2, [3, 4]]

b[2][0] = -100

print(b) # [1, 2, [-100, 4], 100]

"""a is changed because b is filled with references to the same items in a
(except for the newly appended 100 at the end)
"""
print(a) # [1, 2, [-100, 4]]

"""
Shallow copies are separate objects but are filled with references to the same items in the original object.
"""
```

```python
"""
Deep copies

deep copies are fully isolated and separate. The container is a different object and 
all its items are fully new objects as well with the same values as the original objects.

There is no built in deep copy function, but there is the copy module from the standard library
"""

import copy
a = [1, 2, 3]
b = copy.deepcopy(a)

b[0] = -100

print(a) # [1, 2, 3]
print(b) # [-100, 2, 3]

"""
NOTE using deepcopy is expensive and isn't actually recommended unless you really need it.

This also failes if the objects involve system or runtime states like open files, network connections,
threads, generators, and so on
"""
```

```python
"""
Object Representation and Printing

Printing an object or converting it to a string gives it a 'nice' human readable representation
"""
from datetime import date
d = date(2026, 8, 1)
print(d) # 2026-08-01 as a date object

print(str(d)) # 2026-08-01 as a string

# in debugging though, you can't tell if it is a date object or a string

"""
__repr__ is the special method that gives a representation of an object
"""

repr(d) # 'datetime.date(2026, 8, 1)' as a string

print(repr(d)) # datetime.date(2026, 8, 1) as a date object

# !r is the format to call the __repr__ method on an object when using f-strings
print(f'The date is {d!r}')
```
