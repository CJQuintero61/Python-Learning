```python
# interest.py
# calculates compound interest

principal = 1000    # initial amount
rate = 0.05         # interest rate
numyears = 5        # number of years
year = 1

while year <= numyears:
    principal = principal * (1 + rate)
    print(year, principal)
    year = year + 1
```

```python
# interest.py
# calculates compound interest

principal = 1000    # initial amount
rate = 0.05         # interest rate
numyears = 5        # number of years
year = 1

'''
This is also a comment
We will now make the printed block nicer
'''

while year <= numyears:
    principal = principal * (1 + rate)
    print(f'{year:>3d} {principal:0.2f}')
    year = year + 1

'''
The >3d means to right align a 3 digit decimal number
0.2f means round to 2 decimal places
'''
```

## Tuples

Tuples can be extracted with indexing

"""

- indexing
- slicing

  - replace

```python
# Tuples
portfolio = []
with open('portfolio.txt') as file:
    for line in file:
        row = line.split(',')
        name = row[0]
        shares = int(row[1])
        price = float(row[2])
        holding = (name, shares, price)
        portfolio.append(holding)
print(portfolio)
```

```python
# Assuming we did have a file that created the portfolio
# we can then loop over it

portfolio = [
    ('Google',1,2),
    ('Microsoft',3,4),
    ('Tony',5,6),
]

total = 0.0
for name, shares, price in portfolio: # _ could also be used here to discard the name
    total = total + shares * price


# or use list comprehension
# in this case we use _ for the name to discard it since we don't need it
total_two = sum([shares * price for _, shares, price in portfolio])

print(total)
print(total_two)
```

  - it is not recommended though

__len__()
- this is to make the class be able to use the built in len() function in whatever way you define

__repr__()
- changes how the object is printed and displayed
- it is good to always define this for easier debugging

```python
class Stack:
    # the first argument in class methods always refers to the object itself
    def __init__(self): # initialize the stack
        self._items = []
    
    def push(self, item):
        self._items.append(item)
    
    def pop(self):
        return self._items.pop()
    
    def __repr__(self):
        return f'<{type(self).__name__} at 0x{id(self):x}, size={len(self)}>'
    
    def __len__(self):
        return len(self._items)
```

### Inheritance
to inherit another class' methods

```python
class MyStack(Stack):
    def swap(self):
        a = self.pop
        b = self.pop
        self.push(a)
        self.push(b)
```

```python
# You can also redefine inherited methods

class NumericStack(Stack):
    # the push method is redefined
    def push(self, item):
        if not isinstance(item, (int, float)):
            raise TypeError("Only ints or floats allowed")
        # here, super will refer to the prior objects definition of push
        super().push(item)
```

Composition is when you treat another object as an internal detail of a new class
- basically the class has an object as an attribute
