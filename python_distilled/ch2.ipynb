## Operations on Sequences

sequences
- strings
- lists
- tuples

all indexable from [0] onwards

operations
- concatenation
- s * n
  - makes n compies of s, where n is an integer
- s[i] indexing
- s[i:j] slicing
- s[i:j:s] extended slicing
- len(s)

NOte that s * n is a shallow copy
- changes to 1 variable will occur in the shallow copied variable

slicing is hard to understand at times so be careful
- you can also use slice(start, end) and store that slice in a varialbe


```python
# say i wanted to replicate a 4 times so i want to get
# [[1,2,3], [1,2,3], [1,2,3], [1,2,3]]
# if we shallow copy with s *n, changes in 1 variable appear in the other variable
lis = [1,2,3]
x = [lis]
shallow = 4 * x
lis[0] = -7 # will also change what is stored in shallow
print(lis)
print(shallow)

# instead do this, so changes to a aren't reflected in c
a = [1,2,3]
c = [list(a) for _ in range(4)]
a[0] = -7
print(a)
print(c)
```

## Operations on Mutable sequences
strings and tuples are immutable and cannot be modified after creation

operations on mutable sequences
- index assignment s[i] = x
- slice assignment s[i:j] = x
- extended slice assignment s[i:j:stride] = x
- delete an element del s[i]
- delete a slice del s[i:j]
- delete an extended slice del s[i:j:stride]

if doing a slicing assignment, the left and right sides must have the same number of elements

a = [1,2,3,4,5]
a[1::2] = [10,11] # allowed since a = [1,10,3, 11, 5]
a[1::2] = [30, 40, 50] # ValueError since only 2 elemetns in slice on left


these apply to python lists
- things like numpy lists have different rules

## Generator Expressions

A generator expression is an objet that carries out the same computation as a list comprehension
but produces the result iteratively.
- same syntax as list comprehension but use ()

list comprehensions actually make a new list
- generators do not make a new list

You might think then why use these
- this can improve performance and save memory

generators cannot make an object that can be used like a sequence
- no indexing
- no usual list operations

you can use list() to convert a generator expression to a list
clist = list(comments)


```python
nums = [1,2,3,4]
squares = (x*x for x in nums)

print(squares)
print(next(squares))
# 1
print(next(squares))
# 4
for n in squares: print(n)
# 9
# 16

# generators compute their value on demand when you need it

# when the generator is used up, you cannot use it any more
# print(next(squares))
# StopIteration Error
```

```python
# generators can save memory and performance

# read a file
f = open('data.txt')
lines = (t.strip() for t in f)  # read lines, strip trailing and leading whitespace
comments = (t for t in lines if t[0] == '#') # extract comments from the file
for c in comments:
    print(c)

# at no point is the entire file loaded into memory at once
```
